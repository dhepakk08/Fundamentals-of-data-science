```markdown
## Scenario 1: Student Performance Analysis
**Question:** How would you use NumPy arrays to calculate the average score for each subject and determine the subject with the highest average score? Assume a $4 \times 4$ matrix that stores marks of each student in the given order: Math, Science, English, and History.
```

````python
import numpy as np
student_scores = np.array([
    [85, 90, 78, 92],
    [88, 85, 90, 80],
    [92, 88, 84, 85],
    [79, 92, 86, 89]
])

subjects = ['Math', 'Science', 'English', 'History']

# Calculate the average score for each subject (average along columns, axis=0)
average_scores = np.mean(student_scores, axis=0)

# Identify the subject with the highest average score
highest_idx = np.argmax(average_scores)

print("Average scores per subject:")
for sub, avg in zip(subjects, average_scores):
    print(f"  {sub}: {avg:.2f}")

print(f"\nSubject with the highest average score: {subjects[highest_idx]} ({average_scores[highest_idx]:.2f})")
````

```markdown
## Scenario 2: Online Sales Data Analysis
**Question:** How would you find the average price of all the products sold in the past month? Assume a $3 \times 3$ matrix with each row representing the sales for a different product.
```

````python
# Assume a 3x3 matrix where each row contains prices of 3 transactions for that product
product_sales = np.array([
    [19.99, 21.50, 20.00],  # Product A prices
    [49.99, 45.00, 49.99],  # Product B prices
    [9.99,  10.50, 9.50]    # Product C prices
])

# Find the overall average price of all products sold
overall_average = np.mean(product_sales)

# Alternatively, calculate average price per product
avg_per_product = np.mean(product_sales, axis=1)

print(f"Average price of all products sold: ${overall_average:.2f}")
print("Average price per individual product:")
for idx, avg in enumerate(avg_per_product):
    print(f"  Product {chr(65+idx)}: ${avg:.2f}")
````

```markdown
## Scenario 3: Real Estate Analysis
**Question:** Using NumPy arrays and operations, how would you find the average sale price of houses with more than four bedrooms in the neighborhood?
```

````python
# Simulated house_data: [Bedrooms, Square Footage, Sale Price]
house_data = np.array([
    [3, 1500, 250000],
    [5, 3200, 550000],
    [4, 2200, 380000],
    [6, 4000, 720000],
    [5, 2800, 480000],
    [2, 1100, 180000]
])

# Boolean masking to filter houses with > 4 bedrooms (column 0)
bedrooms_mask = house_data[:, 0] > 4

# Extract prices (column 2) for houses that match the mask
prices_above_four_beds = house_data[bedrooms_mask, 2]

# Calculate average price
average_price = np.mean(prices_above_four_beds)

print(f"Number of houses matching criteria: {len(prices_above_four_beds)}")
print(f"Average sale price of houses with more than 4 bedrooms: ${average_price:,.2f}")
````

```markdown
## Scenario 4: Quarterly Sales Performance
**Question:** Using NumPy arrays and arithmetic operations, calculate the total sales for the year and determine the percentage increase in sales from the first quarter to the fourth quarter.
```

````python
# Sales data representing Quarters 1 to 4
sales_data = np.array([120000, 145000, 160000, 195000])

# Total annual sales
total_annual_sales = np.sum(sales_data)

# Percentage increase from Q1 (index 0) to Q4 (index 3)
q1_sales = sales_data[0]
q4_sales = sales_data[3]
percentage_increase = ((q4_sales - q1_sales) / q1_sales) * 100

print(f"Total sales for the year: ${total_annual_sales:,.2f}")
print(f"Percentage increase from Q1 to Q4: {percentage_increase:.2f}%")
````

```markdown
## Scenario 5: Fuel Efficiency Analysis
**Question:** How would you use NumPy arrays and arithmetic operations to calculate the average fuel efficiency and determine the percentage improvement in fuel efficiency between two car models?
```

````python
# fuel_efficiency array (mpg for different car models)
fuel_efficiency = np.array([25.5, 30.2, 18.0, 35.4, 28.1])

# Calculate average fuel efficiency
average_efficiency = np.mean(fuel_efficiency)

# Compare Model index 2 (18.0 mpg) and Model index 3 (35.4 mpg)
model_a = fuel_efficiency[2]
model_b = fuel_efficiency[3]

percentage_improvement = ((model_b - model_a) / model_a) * 100

print(f"Average fuel efficiency of the fleet: {average_efficiency:.2f} MPG")
print(f"Fuel efficiency of Model A: {model_a} MPG")
print(f"Fuel efficiency of Model B: {model_b} MPG")
print(f"Percentage improvement from Model A to Model B: {percentage_improvement:.2f}%")
````
